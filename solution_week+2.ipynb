```python
def load_data():
    try:
        df=pd.read_csv('kc_house_data.csv')
        print("Data loaded successfuly!")
        return df

    except Exception as e:
        print("Failed to load data!")
        print(e)
        return None    
```

```python
price = np.array(df['price'])
sqft = np.array(df['sqft_living'])
```

```python
print("Mean:", np.mean(price))
print("Median:", np.median(price))
print("Std:", np.std(price))
print("Variance:", np.var(price))
```

```python
print("25th percentile:", np.percentile(price, 25))
print("75th percentile:", np.percentile(price, 75))
```

```python
print("Houses above mean sqft_living:", np.sum(sqft > np.mean(sqft)))
```

```python
df['price_per_sqft'] = np.round(df['price'] / df['sqft_living'], 2)
df['total_rooms'] = df['bedrooms'] + df['bathrooms']
df['house_age'] = 2026 - df['yr_built']
df['log_price'] = np.log10(df['price'])

display(df[['price', 'sqft_living', 'price_per_sqft', 'total_rooms', 'house_age', 'log_price']].head())
```

```python
df['is_renovated'] = np.where(df['yr_renovated'] > 0, 'Yes', 'No')
print(df['is_renovated'].value_counts())
```

```python
df['price_tier'] = np.where(df['price'] < 350000, 'Budget',
                   np.where(df['price'] <= 750000, 'Mid-Range', 'Luxury'))
print(df['price_tier'].value_counts())
```

```python
corr = np.corrcoef(df['price'], df['sqft_living'])
print("Correlation matrix:\n", corr)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['sqft_living'], bins=40, edgecolor='black')
plt.axvline(df['sqft_living'].mean(), color='red', linestyle='--',
            label=f"Mean = {df['sqft_living'].mean():.0f}")
plt.title('Distribution of Living Area')
plt.xlabel('sqft_living')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.boxplot(df['bedrooms'])
plt.title('Boxplot of Bedrooms')
plt.ylabel('Bedrooms')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['sqft_living'], df['price'], alpha=0.4)
plt.title('Living Area vs Price')
plt.xlabel('sqft_living')
plt.ylabel('price')
plt.grid(True)
plt.show()
```

```python
grade_price = df.groupby('grade')['price'].mean()
plt.figure(figsize=(8, 5))
plt.bar(grade_price.index, grade_price.values)
plt.title('Average Price by Grade')
plt.xlabel('Grade')
plt.ylabel('Mean price')
plt.grid(True)
plt.show()
```

```python
yearly = df[df['yr_built'] >= 1950].groupby('yr_built')['price'].mean()
plt.figure(figsize=(10, 5))
plt.plot(yearly.index, yearly.values)
plt.title('Average Price by Year Built (1950+)')
plt.xlabel('Year built')
plt.ylabel('Mean price')
plt.grid(True)
plt.show()
```

```python
tier_counts = df['price_tier'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(tier_counts.values, labels=tier_counts.index, autopct='%1.1f%%')
plt.title('Price Tier Distribution')
plt.show()
```
